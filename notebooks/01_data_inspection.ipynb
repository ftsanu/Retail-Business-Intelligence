```python
df_clean = df.copy()

print("Working copy created successfully!")
```

```python
df_clean['Order Date'] = pd.to_datetime(
    df_clean['Order Date'],
    errors='coerce'
)

df_clean['Ship Date'] = pd.to_datetime(
    df_clean['Ship Date'],
    errors='coerce'
)
```

```python
print(df_clean[['Order Date', 'Ship Date']].dtypes)
```

```python
df_clean['Postal Code'] = (
    df_clean['Postal Code'].astype('string')
)
```

```python
print("Missing Order Dates:",
      df_clean['Order Date'].isnull().sum())

print("Missing Ship Dates:",
      df_clean['Ship Date'].isnull().sum())
```

```python
invalid_shipping = (
    df_clean['Ship Date'] < df_clean['Order Date']
).sum()

print("Invalid Shipping Dates:", invalid_shipping)
```

```python
df_clean['Year'] = (
    df_clean['Order Date'].dt.year
)
```

```python
df_clean['Month'] = (
    df_clean['Order Date'].dt.month_name()
)
```

```python
df_clean['Month Number'] = (
    df_clean['Order Date'].dt.month
)
```

```python
df_clean['Shipping Days'] = (
    df_clean['Ship Date'] -
    df_clean['Order Date']
).dt.days
```

```python
df_clean['Profit Margin'] = np.where(
    df_clean['Sales'] != 0,
    (df_clean['Profit'] / df_clean['Sales']) * 100,
    np.nan
)
```

```python
cursor.execute("DESCRIBE retail_orders")

mysql_columns = [
    column[0] for column in cursor.fetchall()
]

python_columns = df_import.columns.tolist()

print("Columns match:", mysql_columns == python_columns)
```

```python
df_import = df_import.where(
    pd.notnull(df_import),
    None
)
```

```python
cursor.execute(
    "SELECT COUNT(*) FROM retail_orders"
)

print("Existing Records:", cursor.fetchone()[0])
```

```python
columns = ", ".join(
    f"`{col}`" for col in df_import.columns
)

placeholders = ", ".join(
    ["%s"] * len(df_import.columns)
)

insert_query = f"""
INSERT INTO retail_orders ({columns})
VALUES ({placeholders})
"""

print(insert_query)
```

```python
data = [
    tuple(row)
    for row in df_import.itertuples(
        index=False,
        name=None
    )
]

print("Total records prepared:", len(data))
```

```python
batch_size = 1000

for i in range(0, len(data), batch_size):

    batch = data[i:i + batch_size]

    cursor.executemany(
        insert_query,
        batch
    )

    connection.commit()

    print(f"Inserted {min(i + batch_size, len(data))} records")
```
